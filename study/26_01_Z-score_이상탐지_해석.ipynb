```python
# 실습1 코드

Mi_n = "../csv/26_mimii_new.csv"
Mi_f = "../csv/26_mimii_features.csv"

df_f = pd.read_csv(Mi_f)

print(df_f.shape) 
print(df_f["label"].value_counts().to_dict())
# {0: 180, 1: 40}  ==> 0 정상이 18개 , 1 비정상이 40개 존재한다는 것을 알 수 있음
```

### 실습 2. 특징값 분포 정상·이상 비교
목표 : 한 특징의 분포를 정상·이상 색으로 나눠 그려 차이를 보기

단계
- 특징 하나를 히스토그램으로 그리기
- 정답으로 색을 나눠 겹쳐 그리기
- 정상·이상 분포가 어디서 갈리는지 관찰

예상 결과
- 이상은 정상보다 값이 큰 쪽으로 분포가 치우침


```python
# 실습2 코드

sns.histplot(data = df_f, x = "rms", hue = "label") 
# hue='label'을 사용하여 각 label별 RMS 값의 분포를 색상으로 구분하고, 클래스 간 RMS 특성의 차이를 비교 가능
# 그래프에서 겹친다는 것은 그 RMS 값만으로는 두 label을 명확하게 구분하기 어렵다는 뜻
# 반대로 두 분포가 많이 떨어져 있다면 rms가 label을 구분하는 데 유용한 feature일 가능성이 높다
plt.title("RMS 분포")
plt.show()


# RMS 분포를 확인한 결과, label 0은 주로 낮은 RMS 값(약 0.03~0.07)에 분포하고, label 1은 상대적으로 높은 RMS 값(약 0.09~0.14)에 분포하는 경향 확인
# 약 0.055~0.08 구간에서는 두 label의 분포가 일부 겹치지만, 
# 전체적으로 두 클래스 간 RMS 분포 차이가 뚜렷하여 RMS가 클래스 구분에 유용한 특성으로 판단된다.

# 겹친다는 것 = 잘못 분류됐다는 뜻 아님, 히스토그램에서 같은 RMS 구간에 label 0과 1이 존재한다는 뜻일 뿐, 
# 다른 feature를 함께 사용하면 그 데이터들을 충분히 구분할 수 있다.
# 두 분포가 달라 보인다고 RMS 하나가 좋은 분류 변수라고 확정할 수도 없다. 
# 현재 데이터에서는 연관성이 있어 보인다는 정도일 뿐, 새로운 데이터에서도 같은 관계가 유지되는지는 검증 필요
```

### 실습 3. 정상 데이터 기준선 세우기
목표 : 정상만 골라 특징별 평균·표준편차로 기준선을 세우기

단계
- 정답이 정상인 행만 골라내기
- 정상 개수를 확인
- 세 특징의 평균과 표준편차를 출력

예상 결과
- 정상 180개, rms 평균 0.05·표준편차 0.012 등


```python
# 실습3 코드

normal = df_f[df_f['label'] == 0]
print('정상 개수:', len(normal))

for f in ['rms', 'spectral_centroid', 'zero_crossing_rate']:
    print(f, round(normal[f].mean(), 3), round(normal[f].std(), 3))
```

### 실습 4. rms Z-score 계산과 최댓값
목표 : rms를 정상 기준으로 Z-score로 바꾸고 가장 튄 값을 찾기

단계
- 정상 평균·표준편차로 Z-score를 계산
- 결과를 새 열로 담기
- Z가 큰 순으로 정렬해 가장 튄 샘플 확인

예상 결과
- 가장 튄 Z가 약 9로 정상에서 크게 벗어남


```python
# 실습4 코드
rms_mean = normal["rms"].mean()
rms_std = normal["rms"].std()

df_f["Z_rms"] = (df_f["rms"] - rms_mean) / rms_std
df_f.head()

top = df_f.sort_values("Z_rms", ascending = False)[["rms", "Z_rms", "label"]].head()
print(top.round(2).to_string(index=False))
```

### 실습 5. 임계값 적용과 2·3 비교
목표 : 절댓값 기준으로 이상을 표시하고 임계값 2와 3 결과를 비교

단계
- Z의 절댓값을 구하기
- 절댓값이 3을 넘으면 이상으로 표시
- 임계값 2와 3의 탐지 개수를 비교

예상 결과
- 임계값 3은 23건, 2는 36건 (낮출수록 많이 잡힘)

```python
# 실습5 코드
df_f["Z_abs"] = df_f["Z_rms"].abs()
# df_f.head()

df_f["is_anomaly"] = df_f["Z_abs"] > 3.0   # is가 들어가면 이게 T/F 값으로 들어간다고 생각하면 돼
df_f["is_anomaly_2"] = df_f["Z_abs"] > 2.0   
# df_f.head()

print('임계값 3:', int(df_f['is_anomaly'].sum()))
print('임계값 2:', int(df_f['is_anomaly_2'].sum()))
```

### 실습 6. 세 특징 종합 판정
목표: 세 특징에 각각 Z-score를 적용하고 하나라도 넘으면 이상으로 종합

단계
- 세 특징마다 정상 기준으로 Z를 계산
- 특징 중 하나라도 절댓값이 3을 넘으면 이상 표시
- rms 단독 결과와 종합 결과를 비교

예상 결과
- 종합 이상은 39건으로 rms 단독 23건보다 많음

```python
# 실습6 코드

# df의 index갯수만큼 False로 채워진 Series를 제작  
is_any = pd.Series(False, index = df_f.index)

for f in ['rms', 'spectral_centroid', 'zero_crossing_rate']:
    z = (df_f[f] - normal[f].mean()) / normal[f].std()
    is_any = is_any | (np.abs(z) > 3.0)

print('종합 이상:', int(is_any.sum()))  # 39
```

### 실습 7. 탐지 결과 정비 문장으로 정리
목표: 가장 튄 이상 샘플을 사람이 읽을 정비 문장으로 옮기기

단계
- 이상으로 표시된 것 중 Z 절댓값 상위 몇 개 고르기
- 각 샘플의 벗어난 방향과 배수를 확인
- 점검 권장 문장으로 출력

예상 결과
- 소리 세기가 표준편차 몇 배 위로 벗어남, 점검 권장 문장

```python
# 실습7 코드
hits = df_f[df_f['is_anomaly']].sort_values('Z_rms', ascending=False).head(3)

for i, row in hits.iterrows():
    direction = '위' if row['Z_rms'] > 0 else '아래'
    print(f"{i}번: 소리 세기가 표준편차 {row['Z_rms']:.1f}배 {direction}로 벗어남, 점검 권장")
```

### 실습 8. Z-score가 놓친 이상 확인
목표: rms 하나로는 놓친 진짜 이상을 찾고 두 특징 산점도로 확인

단계
- 임계값을 넘지 않았는데 실제로는 이상인 샘플을 골라내기
- 놓친 개수를 세기
- 두 특징 산점도에서 정상 속 이상 위치 확인

예상 결과
- rms 단독으로 17건을 놓침 — 한 특징만으로는 부족

```python
#실습8 코드

missed = df_f[(df_f["Z_rms"] <= 3.0) & (df_f["label"] == 1)]
missed.head()

print(f"놓친 이상 : {len(missed)}건") 
# rms 값이 정사이더라도 이상한 값이 label이 1로 잡힌 것으로도 보임
# 같이 봐야함. 어떻게? sctterplot을 통해 이상 구간을 보여주는 것이 좋습니다.

sns.scatterplot(data=df_f, x = "rms", y = "zero_crossing_rate", hue = "label")
plt.title("RMS vs Zero Crossing Rate")
plt.show()

# 3개 N차원으로 보는 것이 좋음
```
